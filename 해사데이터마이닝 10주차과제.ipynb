```python
DATA_DIR = Path('/content/type3')
file_path = DATA_DIR / 'retail_sales_reg1.csv'
```

```python
df = pd.read_csv('/content/retail_sales_reg1.csv')
```

```python
print('[데이터 상위 5행]')
print(df.head(), '\n')
```

```python
print('[기본 정보]')
print(df.info(), '\n')
```

```python
model = smf.ols('sales ~ ad_cost + staff + event_cnt', data=df).fit()
```

```python
print('[회귀 요약 결과]')
print(model.summary())
```

```python
print('\n[회귀계수]')
print(model.params)
```

```python
print('\n[p-value]')
print(model.pvalues)
```

```python
print('\n[R-squared]')
print(model.rsquared)
```

```python
beta_ad = model.params['ad_cost']
print('\n광고비 10단위 증가 시 예상 매출 변화:', beta_ad * 10)

```

```python
sig_vars = model.pvalues[model.pvalues < 0.05].index.tolist()
print('\n유의한 변수:', sig_vars)
```

```python
from google.colab import drive
from pathlib import Path
import pandas as pd
import numpy as np
import statsmodels.formula.api as smf
from sklearn.metrics import mean_squared_error
```

```python
drive.mount('/content/drive')

```

```python
DATA_DIR = Path('/content/type3')
file_path = DATA_DIR / 'weather_reg2.csv'
```

```python
df = pd.read_csv('/content/weather_reg2.csv')
```

```python
print('[데이터 상위 5행]')
print(df.head(), '\n')
```

```python
print('[기본 정보]')
print(df.info(), '\n')
```

```python
model = smf.ols('temperature ~ solar + wind + o3', data=df).fit()
```

```python
print('[회귀 요약 결과]')
print(model.summary())
```

```python
print('\n[회귀계수]')
print(model.params)
```

```python
print('\n[p-value]')
print(model.pvalues)

```

```python
print('\n[R-squared]')
print(model.rsquared)
```

```python
new_df = pd.DataFrame({
    'solar': [100],
    'wind': [5],
    'o3': [30]
})
```

```python
pred = model.predict(new_df)
print('\n[예측값]')
print(pred)
```

```python
pred_frame = model.get_prediction(new_df).summary_frame(alpha=0.05)
print('\n[예측 요약표]')
print(pred_frame)
```

```python
y_true = df['temperature']
y_pred = model.predict(df)
```

```python
mse = mean_squared_error(y_true, y_pred)
rmse = np.sqrt(mse)
```

```python
print('\n[MSE / RMSE]')
print('MSE :', mse)
print('RMSE:', rmse)
```
